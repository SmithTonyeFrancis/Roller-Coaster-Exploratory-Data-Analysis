```python
#The next step is to remove columns that we do not want and leave out only the columns
#that we want to keep
#Also the copy() function creates an independent duplicate of a DataFrame. It gives us a new isolated copy

df=df[['coaster_name', 
    #'Length', 'Speed', 
    'Location', 'Status', 
    #'Opening date',
    # 'Type', 
    'Manufacturer', 
    #'Height restriction', 'Model', 'Height',
    #'Inversions', 'Lift/launch system', 'Cost', 'Trains', 'Park section',
    #'Duration', 'Capacity', 'G-force', 'Designer', 'Max vertical angle',
    #'Drop', 'Soft opening date', 'Fast Lane available', 'Replaced',
    #'Track layout', 'Fastrack available', 'Soft opening date.1',
    #'Closing date', 
    #'Opened', 
    #'Replaced by', 'Website',
    #'Flash Pass Available', 'Must transfer from wheelchair', 'Theme',
    #'Single rider line available', 'Restraint Style',
    #'Flash Pass available', 'Acceleration', 'Restraints', 'Name',
       'year_introduced', 'latitude', 'longitude', 'Type_Main',
       'opening_date_clean', 
    #'speed1', 'speed2', 'speed1_value', 'speed1_unit',
       'speed_mph', 
    #'height_value', 'height_unit', 
    'height_ft',
       'Inversions_clean', 'Gforce_clean']].copy()
    
```

```python
#the Opening_date_clean column was showing as an object
#So we need to change it to a datetime datatype

df['opening_date_clean']=pd.to_datetime(df['opening_date_clean'])
```

```python
#Step 3: Feature Understanding 
#1. Plotting Feature Distribution
#1.1 Histogram
#1.2 KDE
#1.3 Boxplot
```

```python
#The value_counts function tells us the number of times each unique value
#appears in a column

df['Year_Introduced'].value_counts()
```

```python
#Plotting the Top years coasters introduced for the top 10 rows or years

df['Year_Introduced'].value_counts()\
    .head(10).plot(kind='bar', title='Top 10 Years Coasters Introduced')
```

```python
#Saving the plot as a matplotlib file

ax=df['Year_Introduced'].value_counts()\
    .head(10).plot(kind='bar', title='Top 10 Years Coasters Introduced')
ax.set_xlabel('Year Introduced')
ax.set_ylabel('Count')
```

```python
#here we are checking for the distribution of the different roller coaster speed.
#Checking out the values are spread out and the underlying structure they form

ax=df['Speed_mph'].plot(kind='hist', bins=20, title='Coaster Speed (mph)')
ax.set_xlabel('Speed (mph)')
plt.show()
```

```python
ax=df['Speed_mph'].plot(kind='kde',title='Coaster Speed (mph)')
ax.set_xlabel('Speed (mph)')
plt.show()
```

```python
#Step 4: Feature Relationships
#4.1 Scatterplot
#4.2 Heatmap Correlation
#4.3 Pairplot
#4.4 Groupby comparisons
```

```python
#So we are comparing how the speed of each roller coaster function according to height

df.plot(kind='scatter', x='Speed_mph', y='Height_ft', title = 'Coaster Speed vs Height')
plt.show()
```

```python
#Using seaborn to plot a better and more advanced graph plot

sns.scatterplot( x='Speed_mph', y='Height_ft', hue='Year_Introduced',data=df)
plt.show()
```

```python
#Comparing multiple features against each other using the pairplot function in seaborn

sns.pairplot(df, vars=['Year_Introduced', 'Speed_mph', 'Height_ft', 
                       'Inversions_Clean', 'Gforce'],
            hue='Type_Main')
plt.show()
```

```python
#Checking the correlation between columns that are numeric

df_corr=df[['Year_Introduced', 'Speed_mph', 'Height_ft', 'Inversions_Clean', 'Gforce']]\
.dropna().corr()
```

```python
#Using seaborns heatmap to check the correlation between the data

sns.heatmap(df_corr, annot=True)
plt.show()
```

```python
#Question 1
#What are the locations with the fastest rollercoasters (minimum of 10)?

ax=df.query('Location != "Other"')\
    .groupby('Location')['Speed_mph']\
    .agg(['mean', 'count'])\
    .query('count>=10')\
    .sort_values('mean')['mean']\
    .plot(kind='barh', figsize=(12,5), title ='Average Coaster Speed by Location')
ax.set_xlabel('Average Coaster Speed')

plt.show()
```
